## Exercise 2.4 — Convert rows to SensorThings Observations
Set the Datastream ID and create payloads for temperature.

```python
TEMPERATURE_DATASTREAM_ID = None  # TODO: replace

def row_to_observation(row):
    return {
        'phenomenonTime': row['phenomenonTime'],
        'result': float(row['temperature_C']),
        'Datastream': {'@iot.id': TEMPERATURE_DATASTREAM_ID},
        'parameters': {'quality_flag': row.get('quality_flag', 'raw'), 'source_file': 'observations.csv'}
    }

example_payload = row_to_observation(df_csv.iloc[0])
example_payload
```
